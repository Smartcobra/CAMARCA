### Reading the accuracy / loss curves

`advanced_evaluation.py` splits cases into **train / val / test** (default **60% / 20% / 20%**).

`accuracy_curve.png` and `loss_curve.png` plot **running** top-1 accuracy and rank loss from the **val** and **test** splits only (not train, not all `--max-cases` in one line).

**Example:** `--max-cases 45` → about **27 train, 9 val, 9 test** (exact counts depend on integer rounding). The x-axis length matches **the number of val (or test) cases** — so you typically see about **9** points per line, **not 45**. Train cases are used for weight tuning and the sklearn reranker, but they are not shown on these two running curves.

```python
# Plot preview
from PIL import Image
from IPython.display import display
for img in [
    'outputs/evaluation/accuracy_curve.png',
    'outputs/evaluation/loss_curve.png',
    'outputs/evaluation/advanced_metrics.png',
    'outputs/evaluation/confusion_matrix.png',
]:
    p = Path(img)
    if p.exists():
        print('\n', img)
        display(Image.open(p))
    else:
        print('\nMissing:', img)
```
